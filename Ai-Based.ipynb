```python
import pandas as pd

# Load dataset
df = pd.read_csv("Train.csv")

# Show columns
print("Columns are:")
print(df.columns)

# Show first 5 rows
df.head()
```

```python
# Convert date_time
df['date_time'] = pd.to_datetime(df['date_time'])

# Extract features
df['hour'] = df['date_time'].dt.hour
df['day'] = df['date_time'].dt.day
df['month'] = df['date_time'].dt.month

df.head()
```

```python
df['co2_emission'] = df['traffic_volume'] * 0.5 + df['air_pollution_index'] * 0.5
```

```python
import matplotlib.pyplot as plt

#  STEP 1: Sort FIRST
df = df.sort_values(by='date_time')
df = df.reset_index(drop=True)

#  STEP 2: Traffic (clean visualization)
plt.figure(figsize=(12,5))
plt.scatter(df['date_time'], df['traffic_volume'], s=1)
plt.title("Traffic Level Over Time (Clean)")
plt.xlabel("Time")
plt.ylabel("Traffic Volume")
plt.show()

#  STEP 3: AQI
plt.figure(figsize=(12,5))
plt.scatter(df['date_time'], df['air_pollution_index'], s=1)
plt.title("AQI Over Time (Clean)")
plt.xlabel("Time")
plt.ylabel("AQI")
plt.show()

# ✅ STEP 4: CO2
plt.figure(figsize=(12,5))
plt.scatter(df['date_time'], df['co2_emission'], s=1)
plt.title("CO2 Emission Over Time (Clean)")
plt.xlabel("Time")
plt.ylabel("CO2")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor

# Features
X = df[['hour', 'day', 'month']]

# Target
y = df['traffic_volume']

# Split (time-based)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# Model
model = RandomForestRegressor()
model.fit(X_train, y_train)

# Prediction
y_pred = model.predict(X_test)
```

```python
import matplotlib.pyplot as plt

# Get correct time values for test data
time_test = df['date_time'].iloc[len(X_train):]

plt.figure(figsize=(12,5))

plt.plot(time_test, y_test.values, label="Actual")
plt.plot(time_test, y_pred, label="Predicted")

plt.legend()
plt.title("Traffic Prediction (Actual vs Predicted)")
plt.xlabel("Time")
plt.ylabel("Traffic Volume")

plt.show()
```

```python
# AQI prediction
y_aqi = df['air_pollution_index']

X_train, X_test, y_train_aqi, y_test_aqi = train_test_split(X, y_aqi, test_size=0.2, shuffle=False)

model_aqi = RandomForestRegressor()
model_aqi.fit(X_train, y_train_aqi)

y_pred_aqi = model_aqi.predict(X_test)
```

```python
# Plot AQI

time_test = df['date_time'].iloc[len(X_train):]

plt.figure(figsize=(12,5))
plt.plot(time_test, y_test_aqi.values, label="Actual AQI")
plt.plot(time_test, y_pred_aqi, label="Predicted AQI")
plt.legend()
plt.title("AQI Prediction")
plt.xlabel("Time")
plt.ylabel("AQI")
plt.show()
```

```python
# CO2 prediction
y_co2 = df['co2_emission']

X_train, X_test, y_train_co2, y_test_co2 = train_test_split(X, y_co2, test_size=0.2, shuffle=False)

model_co2 = RandomForestRegressor()
model_co2.fit(X_train, y_train_co2)

y_pred_co2 = model_co2.predict(X_test)
```

```python
# Plot CO2

plt.figure(figsize=(12,5))
plt.plot(time_test, y_test_co2.values, label="Actual CO2")
plt.plot(time_test, y_pred_co2, label="Predicted CO2")
plt.legend()
plt.title("CO2 Prediction")
plt.xlabel("Time")
plt.ylabel("CO2")
plt.show()
```

```python
# Error Comparison

from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# 🔹 Traffic
mae_traffic = mean_absolute_error(y_test, y_pred)
rmse_traffic = np.sqrt(mean_squared_error(y_test, y_pred))

# 🔹 AQI
mae_aqi = mean_absolute_error(y_test_aqi, y_pred_aqi)
rmse_aqi = np.sqrt(mean_squared_error(y_test_aqi, y_pred_aqi))

# 🔹 CO2
mae_co2 = mean_absolute_error(y_test_co2, y_pred_co2)
rmse_co2 = np.sqrt(mean_squared_error(y_test_co2, y_pred_co2))

# 🔹 Print Results
print("Traffic → MAE:", mae_traffic, "| RMSE:", rmse_traffic)
print("AQI     → MAE:", mae_aqi, "| RMSE:", rmse_aqi)
print("CO2     → MAE:", mae_co2, "| RMSE:", rmse_co2)
```

```python
# Showing all 3 (Traffic, AQI, CO2) in one place

plt.figure(figsize=(12,5))

# Correct time for test data
time_test = df['date_time'].iloc[len(X_train):]

# Traffic
plt.plot(time_test, y_test.values, label="Traffic Actual")
plt.plot(time_test, y_pred, label="Traffic Predicted")

# AQI
plt.plot(time_test, y_test_aqi.values, label="AQI Actual")
plt.plot(time_test, y_pred_aqi, label="AQI Predicted")

plt.legend()
plt.title("Traffic & AQI Prediction Comparison")
plt.xlabel("Time")
plt.ylabel("Values")

plt.show()


plt.figure(figsize=(12,5))

time_test = df['date_time'].iloc[len(X_train):]

# Traffic
plt.plot(time_test, y_test.values, label="Traffic Actual")
plt.plot(time_test, y_pred, label="Traffic Predicted")

# AQI
plt.plot(time_test, y_test_aqi.values, label="AQI Actual")
plt.plot(time_test, y_pred_aqi, label="AQI Predicted")

# CO2
plt.plot(time_test, y_test_co2.values, label="CO2 Actual")
plt.plot(time_test, y_pred_co2, label="CO2 Predicted")

plt.legend()
plt.title("Traffic, AQI & CO2 Prediction Comparison")
plt.xlabel("Time")
plt.ylabel("Values")

plt.xticks(rotation=45)
plt.show()
```

```python
#  Add Error Bar Graph

import numpy as np

labels = ['Traffic', 'AQI', 'CO2']
mae = [mae_traffic, mae_aqi, mae_co2]
rmse = [rmse_traffic, rmse_aqi, rmse_co2]

x = np.arange(len(labels))

plt.figure(figsize=(8,5))
plt.bar(x - 0.2, mae, 0.4, label='MAE')
plt.bar(x + 0.2, rmse, 0.4, label='RMSE')

plt.xticks(x, labels)
plt.legend()
plt.title("Error Comparison (MAE vs RMSE)")
plt.show()
```

```python
#  Route Optimization

import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()

# Add more realistic roads (nodes = locations)
G.add_edge("A", "B", weight=4)
G.add_edge("A", "C", weight=2)
G.add_edge("B", "C", weight=1)
G.add_edge("B", "D", weight=5)
G.add_edge("C", "D", weight=8)
G.add_edge("C", "E", weight=10)
G.add_edge("D", "E", weight=2)
G.add_edge("D", "F", weight=6)
G.add_edge("E", "F", weight=3)

# Find shortest path (least traffic)
shortest_path = nx.shortest_path(G, "A", "F", weight='weight')
shortest_length = nx.shortest_path_length(G, "A", "F", weight='weight')

print("Optimal Traffic-Free Route:", shortest_path)
print("Total Traffic Cost:", shortest_length)
```

```python
pos = nx.spring_layout(G)

# Draw full graph
nx.draw(G, pos, with_labels=True, node_color='lightblue', node_size=2000)

# Draw edge labels (traffic weights)
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

# Highlight shortest path
path_edges = list(zip(shortest_path, shortest_path[1:]))
nx.draw_networkx_edges(G, pos, edgelist=path_edges, edge_color='red', width=3)

plt.title("Traffic-Free Route Optimization")
plt.show()
```
